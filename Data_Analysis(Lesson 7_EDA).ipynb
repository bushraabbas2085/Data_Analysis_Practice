**Lesson 7 Part 2:** Mean, median, standard deviation, distributions & correlation

The mean gets pulled upward by the extremely high salary.
The median is much less affected by that unusual value.
That's why a Data Analyst needs to understand both.

```python
##Mean Vs Median
print("Math Mean:", df["Math"].mean())
print("Math Median:", df["Math"].median())

print("English Mean:", df["English"].mean())
print("English Median:", df["English"].median())

print("Science Mean:", df["Science"].mean())
print("Science Median:", df["Science"].median())
```

**Mean** is the overall average of a column, while **median** is the middle value when the data is arranged in order. The mean is more affected by sudden/extreme changes (outliers), while the median is less affected by them.

**Mean =** good for normally distributed data without major outliers.

**Median =** often better when the data contains extreme values/outliers.

```python
##Standard Deviation
print("Math Standard Deviation:", df["Math"].std())
print("English Standard Deviation:", df["English"].std())
print("Science Standard Deviation:", df["Science"].std())
```

| Standard Deviation | Meaning                         |
| ------------------ | ------------------------------- |
| Small              | Values are close to the mean    |
| Large              | Values are spread out           |
| 0                  | All values are exactly the same |


"English has the highest standard deviation, indicating that English performance varies more among the students."

```python
##Median and Correlation
print("Median : " , df["Math"].median())
print("\nCorrelation")
df[['Math' , 'English' , 'Science']].corr()
```

**Correlation** tells us whether two numerical variables are related.

| Correlation | Meaning                      |
| ----------: | ---------------------------- |
|      **+1** | Perfect positive correlation |
|    **+0.7** | Strong positive relationship |
|    **+0.3** | Weak positive relationship   |
|       **0** | No linear correlation        |
|    **-0.3** | Weak negative relationship   |
|    **-0.7** | Strong negative relationship |
|      **-1** | Perfect negative correlation |

```python
print("Math vs Science:", df["Math"].corr(df["Science"]))

print("Math vs English:", df["Math"].corr(df["English"]))

print("English vs Science:", df["English"].corr(df["Science"]))
```

```python
##Correlation Matrix
df[['Math' , 'English' , 'Science']].corr()
```

The matrix is symmetrical.For example:

Math ↔ Science = 0.851
Science ↔ Math = 0.851

```python
##Correlation Heatmap
corr = df[['Math' , 'English' , 'Science']].corr()
sns.heatmap(corr , annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

**Identifying Outliers**


An outlier is a signal to investigate, not something you automatically delete.

```python
##Boxplot
sns.boxplot(data=df[["Math", "English", "Science"]])

plt.title("Subject Scores - Outlier Detection")
plt.show()
```

```python
##Interquartile Range
for column in ["Math", "English", "Science"]:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print("\n", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("\nOutliers:")
    print(outliers[["Name", column]])
```

**Group Analysis**

Compare groups/categories in a dataset.

```python
##Group by Gender
print(df.groupby("Gender")[["Math", "English", "Science"]].mean())
```

"In this dataset, female students have higher average scores than male students across all three subjects."

```python
print(df[["Math", "English", "Science"]].mean())
```

"Among the three subjects, Science has the highest overall average score, while English has the lowest average score."

```python
print("Highest Math:")
print(df.loc[df["Math"].idxmax(), ["Name", "Math"]])

print("\nLowest Math:")
print(df.loc[df["Math"].idxmin(), ["Name", "Math"]])

print("Highest English:")
print(df.loc[df["English"].idxmax(), ["Name", "English"]])

print("\nLowest English:")
print(df.loc[df["English"].idxmin(), ["Name", "English"]])

print("Highest Science:")
print(df.loc[df["Science"].idxmax(), ["Name", "Science"]])

print("\nLowest Science:")
print(df.loc[df["Science"].idxmin(), ["Name", "Science"]])
```

Ayesha achieved the highest Math score (95), while Ahmed had the lowest (76). Hina achieved the highest English score (94), while Usman had the lowest (75). Sara achieved the highest Science score (95), while Ahmed had the lowest (79).

Ahmed has the lowest score in both Math and Science.

```python
df["Total"] = df["Math"] + df["English"] + df["Science"]

df["Average"] = df[["Math", "English", "Science"]].mean(axis=1)

print(df[["Name", "Total", "Average"]])

print(df.loc[df["Total"].idxmax(), ["Name", "Total", "Average"]])

print(df.loc[df["Total"].idxmin(), ["Name", "Total", "Average"]])
```

```python
ranking = df.sort_values("Average", ascending=False)

print(ranking[["Name", "Math", "English", "Science", "Total", "Average"]])
```
