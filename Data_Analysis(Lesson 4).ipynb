 **Grouping and Aggregation**

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman", "Hina",
             "Bilal", "Maham", "Hamza", "Zainab"],

    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female"],

    "City": ["Lahore", "Multan", "Lahore", "Multan", "Lahore", "Multan",
             "Lahore", "Multan", "Lahore", "Multan"],

    "Department": ["Computer Science", "Computer Science",
                   "Computer Science", "Computer Science",
                   "Computer Science", "Computer Science",
                   "Software Engineering", "Software Engineering",
                   "Software Engineering", "Software Engineering"],

    "Math": [85, 92, 78, 95, 88, 91, 90, 87, 82, 94],

    "English": [80, 89, 75, 93, 84, 90, 88, 85, 79, 92],

    "Science": [82, 91, 80, 94, 86, 89, 92, 88, 81, 95]
}

df = pd.DataFrame(data)

print(df)
```

```python
##Group by Department
print("Average Math Scores for each Department")
print(df.groupby("Department")["Math"].mean())
```

```python
##Group by Gender
print("Average Math Scores for each Department")
print(df.groupby("Gender")["Math"].mean())
print("\nAverage English Scores for each Department")
print(df.groupby("Gender")["English"].mean())
print("\nAverage Science Scores for each Department")
print(df.groupby("Gender")["Science"].mean())
```

```python
##Aggregation
print(df.groupby("Department")["Math"].agg(["mean" , "max" , "min"]))
```

```python
##Total Math Scores by Department
print("Total Math Scores of each Department")
print(df.groupby("Department")["Math"].sum())
print("\nTotal English Scores of each Department")
print(df.groupby("Department")["English"].sum())
print("\nTotal Science of each Department")
print(df.groupby("Department")["Science"].sum())
```

```python
##Count Student
print("Total students in each Department")
print(df.groupby("Department")["Name"].count())
print("\nTotal students in each City")
print(df.groupby("City")["Name"].count())
```

```python
##Math scores by each city
print("Average Math scores of each City")
print(df.groupby("City")["Math"].mean())
print("\nMax Math scores by each City")
print(df.groupby("City")["Math"].max())
```

```python
print("Average of Math, English, and Science for each department")
print(df.groupby("Department")[["Math","English","Science"]].mean())
```

```python
print("Average total scores of each Department")
df["Total"] = df["Math"] + df["English"] + df["Science"]
print(df.groupby("Department")["Total"].mean())
```

```python
##Creating Summary Tables
print("Avg Performance of each department")
summary = df.groupby("Department")[["Math","English","Science"]].mean()
print(summary)

print("\nAvg Performance")
summary = df.groupby("Department").agg(
        Students = ("Name" , "count"),
        Avg_Math = ("Math" , "mean") ,
        Avg_English = ("English" , "mean"),
        Avg_Science = ("Science" , "mean")
)
print(summary)
```

Software Engineering has more average Math and Science scores.Computer Science has 6 students and Software Engineering has 4 students . Software Engineering performed better overall .

```python
df["Total"] = df["Math"] + df["English"] + df["Science"]
df["Average_Total"] = df["Total"] / 3
print(df)

print("\nOverall department performance")
print(df.groupby("Department")["Average_Total"].mean())
```

```python
df["Total"] = df["Math"] + df["English"] + df["Science"]

analysis = df.groupby("Department").agg(
    Students=("Name", "count"),
    Average_Math=("Math", "mean"),
    Average_English=("English", "mean"),
    Average_Science=("Science", "mean"),
    Average_Total=("Total", "mean")
)

print(analysis)
```
