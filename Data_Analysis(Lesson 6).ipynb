**Data Visualizations** with Matplotlib : Line chart , bar chart , pie chart and scatter plot

```python
import matplotlib.pyplot as plt
```

Line Charts are used for **analyzing or Visualizing Growth**

```python
##Line Charts for analyzing or Visualizing Growth(Best for trends/change)
days = [1, 2, 3, 4, 5]
sales = [10, 15, 12, 20, 25]

plt.plot(days , sales)
plt.title("Sales over 5 Days")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
revenue = [20, 25, 22, 30, 35]

plt.plot(months , revenue)
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

Bar Charts are best for **comparing categories.**

```python
##Bar Charts for Comparing Categories(Best for comparison)
subjects = ["Math", "Science", "English", "Computer"]
marks = [90, 85, 88, 95]

plt.bar(subjects , marks)
plt.title("Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

A pie chart is useful when we want to show how **a whole is divided into different parts**.

```python
##Pie Charts (best for Parts of a Whole)
subjects = ["Math", "Science", "English", "Computer"]
marks = [90, 85, 88, 95]
plt.pie(marks , labels = subjects)
plt.title("Marks Distribution by Subject")
plt.show()
```

A scatter plot is used to examine the **relationship between two numerical variables**.

**Remember:** a scatter plot can show a relationship, but it doesn't by itself prove causation.

```python
hours = [1, 2, 3, 4, 5, 6]
marks = [50, 55, 65, 70, 80, 90]
plt.scatter(hours , marks)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

**Customizing Charts** : Adding grid ,Changing figuresize and Adding legend

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
revenue = [20, 25, 22, 30, 35]

plt.figure(figsize = (8,5))

plt.plot(months , revenue)
plt.title("Monthly Revenue Report")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid()

plt.show()
```

**Pandas + Matplotlib**

```python
import pandas as pd
import matplotlib.pyplot as plt
data = { "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman", "Hina"],
        "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
         "Math": [85, 92, 78, 95, 88, 90],
         "English": [80, 89, 75, 91, 84, 93],
         "Science": [82, 94, 80, 96, 86, 91] }
df = pd.DataFrame(data)
df.plot(x= "Name" , y= "Science" , kind= "bar")
plt.title("Science Marks by Student")
plt.xlabel("Student")
plt.ylabel("Science Marks")
plt.show()
```

Multiple column comparison

```python
df.plot( x="Name", y=["Math", "English", "Science"], kind="bar" )
plt.title("Student Subject Comparison")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

Comparing average of each subject

```python
averages = df[["Math", "English", "Science"]].mean()

print(averages)
averages.plot(kind="bar")

plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.show()
```

# **Mini Project**

```python
averages = df[["Math" , "English" , "Science"]].mean()
print(averages)

averages.plot(kind="bar")
plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.show()
```

Science subject has the highest average

**Math and Science Relationship**

```python
plt.scatter(df["Math"] , df["Science"])
plt.title("Math vs Science Marks")
plt.xlabel("Math Marks")
plt.ylabel("Science Marks")
plt.show()
```
