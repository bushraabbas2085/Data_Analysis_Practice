```python
import pandas as pd
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Hamza",
             "Fatima", "Usman", "Zainab", "Bilal", "Hina"],

    "Age": [20, 21, 22, 20, 23, 21, 22, 20, 24, 21],

    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],

    "Math": [35, 54, 82, 96, 43, 75, 53, 51, 92, 67],

    "English": [75, 90, 65, 95, 80, 92, 70, 88, 60, 94],

    "Science": [68, 85, 60, 92, 75, 95, 65, 82, 55, 90]
}

df = pd.DataFrame(data)
print(df)
```

```python
print("High Scorers : \n" , df[(df['Math']>=90) & (df['Science']>=85)])
```

```python
females = df[df['Gender'] == 'Female']
print('Female High Achiever : \n' , females[(females["Math"]>90) | (females["English"]>90)])
```

```python
print("Top 3 in Math: \n" , df.sort_values("Math" , ascending = False).head(3)[["Name", "Math"]])
```

```python
df["Total"] = df["Math"] + df["English"] + df["Science"]
print("High Scoring Females: \n")
females = df[df['Gender']=="Female"]
print(females.loc[females['Total'].idxmax()])
```
