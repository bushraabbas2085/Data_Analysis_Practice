```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", " Ahmed ", "Ayesha", "Ali", "Usman", "Hina", "Zain"],
    "Age": [20, 21, 150, None, 20, 22, 19, 14],
    "Gender": ["Male", "female", " MALE ", "Female", "Male", "M", "F", "male"],
    "City": ["Lahore", "multan", "LAHORE", "Multan ", "Lahore", "lahore", "MULTAN", "Multan"],
    "Math": [85, 92, 78, None, 85, 95, 88, 500],
    "Attendance": [90, 95, 88, 92, 90, 97, 85, 105]
}

df = pd.DataFrame(data)

print(df)
```

```python
##Data Inspection
print(df.shape)
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.info())
```

```python
##Remove Extra Space from Names
df["Name"] = df["Name"].str.strip()
print(df)
```

```python
##Making the Gender values consistent.
df["Gender"]= df["Gender"].str.strip()
df["Gender"]= df["Gender"].str.lower()
df["Gender"]= df["Gender"].replace(
    {
        "m":"male",
        "f":"female"
    }
)
df["Gender"]= df["Gender"].str.title()
print(df , "\n" ,df["Gender"].unique())
```

```python
##Making the City values consistent.
df["City"]= df["City"].str.strip()
df["City"]= df["City"].str.lower()
df["City"]= df["City"].str.title()
print(df , "\n" ,df["City"].unique())
```

```python
##Find and Remove Duplicates
print(df[df.duplicated()])
df=df.drop_duplicates()
print("\n" , df)
```

```python
##Handle Age outliers
df.loc[(df["Age"]<15) | (df["Age"]>30) , "Age"] =None
print("\n" ,df)
df["Age"] = df["Age"].fillna(df["Age"].median())
print("\n" ,df)
```

```python
##Handle Math Outliers
df.loc[(df["Math"]<0) | (df["Math"]>100) , "Math"] = None
print("\n" , df)
```

```python
##Handle Attendance Outliers
df.loc[(df["Attendance"]<0) | (df["Attendance"]>100) , "Attendance"] = None
print("\n" , df)
```

```python
print(df)
print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```
