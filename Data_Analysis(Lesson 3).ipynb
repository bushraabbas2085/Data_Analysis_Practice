# **Data Cleaning**

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Ali", "Usman"],
    "Age": [20, 21, None, 20, 20, 22],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Male"],
    "Math": [85, 92, 78, None, 85, 95],
    "City": ["Lahore", "Multan", "Lahore", "Multan", "Lahore", "Multan"]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
print(df[df.isnull().any(axis = 1)])
```

```python
clean_df = df.dropna()
print(clean_df)
```

```python
avg_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(avg_age)
print(df)
```

```python
avg_math = df['Math'].mean()
df['Math'] = df['Math'].fillna(avg_math)
print(df)
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman", "Hina"],
    "Gender": ["Male", "female", " MALE ", "Female", "M", "F"],
    "City": ["Lahore", "multan", "LAHORE", "Multan ", "lahore", "MULTAN"],
    "Department": ["CS", "Computer Science", "cs", "CS ", "Computer Science", "CS"]
}

df = pd.DataFrame(data)

print(df)
```

```python
##Inspect Data
print(df['Gender'].unique())
print(df['City'].unique())
print(df['Department'].unique())
```

```python
##Remove Spaces
df['Gender'] = df['Gender'].str.strip()
df['City'] = df['City'].str.strip()
df['Department'] = df['Department'].str.strip()

print(df)
```

```python
##Convert to Lowercase
df['Gender'] = df['Gender'].str.lower()
df['City'] = df['City'].str.lower()
df['Department'] = df['Department'].str.lower()

print(df)
print("\n" , df['City'].unique())
```

```python
##Standardize Values
df['Gender'] = df['Gender'].replace(
    {
        "m":"male",
        "f":"female"
    }
)

df['Department'] =df['Department'].replace(
    {
        "cs":"computer science"
    }
)

print(df)
```

```python
##Format Data
df['Name'] = df['Name'].str.title()
df['City'] = df['City'].str.title()
df['Gender'] = df['Gender'].str.title()
df['Department'] = df['Department'].str.title()

print(df)
```

```python
print(df["Gender"].unique())
print(df["City"].unique())
print(df["Department"].unique())
```

```python
print(df.info())
```

```python
##Handling Outliers and Incorrect values
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman", "Hina", "Zain"],
    "Age": [20, 21, 19, 20, 22, 21, 150],
    "Math": [85, 92, 78, 88, 95, 82, 500],
    "Attendance": [90, 95, 88, 92, 97, 85, 105]
}

df = pd.DataFrame(data)

print(df)
```

```python
##Indicating Outliers
print(df.describe())
print("\nAge Outlier\n", df[(df["Age"]>30) | (df["Age"]<15)])
print("\nMath Score Outlier\n", df[(df["Math"]<0) | (df["Math"]>100)])
print("\nAttendance Outlier\n", df[(df["Attendance"]<0) | (df["Attendance"]>100)])
```

```python
##handle Age outliers
df.loc[(df["Age"]>30) | (df["Age"]<15) , "Age"] =None
print("\n" ,df)
df["Age"] = df["Age"].fillna(df["Age"].median())
print("\n" ,df)
```

```python
##handle Attendance Outlier
df.loc[df["Attendance"]>100 , "Attendance"]=None
print("\n",df)

##handle Math Outlier
df.loc[df["Math"]>100 , "Math"] = None
print('\n',df)
```
