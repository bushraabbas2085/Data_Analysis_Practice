```python
import pandas as pd
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Hamza",
             "Fatima", "Usman", "Zainab", "Bilal", "Hina"],

    "Age": [20, 21, 22, 20, 23, 21, 22, 20, 24, 21],

    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],

    "Study_Hours": [3, 5, 2, 6, 4, 7, 3, 5, 2, 6],

    "Attendance": [75, 90, 65, 95, 80, 92, 70, 88, 60, 94],

    "Score": [68, 85, 60, 92, 75, 95, 65, 82, 55, 90]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.info()
```

```python
print("Average Age :" , df["Age"].mean())
print("Average Study Hours :", df["Study_Hours"].mean())
print("Average Attendance :" , df["Attendance"].mean())
print("Average Scores :" , df["Score"].mean())
print("Maximum Scores :" , df["Score"].max())
print("Minimum Scores :" , df["Score"].min())
```

```python
print(df[df["Score"]>80])
print(df[df["Attendance"]>90])
print(df[df["Study_Hours"]>=5])
print(df[df["Score"]<70])
```

```python
high = df["Score"].idxmax()
print("Highest Scorer :" , df.loc[high])

low = df["Score"].idxmin()
print("\nlowest scorer: " , df.loc[low])

print("\nMost Punctual: " , df.loc[df["Attendance"].idxmax()])

print("\nMost Hardworking: " , df.loc[df["Study_Hours"].idxmax()])
```

# **Insight**

Students who study more hours generally achieve higher scores
