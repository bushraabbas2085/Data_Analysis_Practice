```python
##Load Data
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman", "Hina",
             "Bilal", "Zainab", "Hamza", "Maham"],

    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female"],

    "Math": [85, 92, 78, 95, 88, 90, 76, 89, 94, 91],

    "English": [80, 88, 75, 91, 84, 93, 79, 87, 90, 89],

    "Science": [90, 95, 82, 94, 86, 91, 80, 92, 96, 88]
}

df = pd.DataFrame(data)

print(df)
```

🧮 **Step 3 — Create Total and Average**

```python
df["Total"] = df["Math"] + df["English"] + df["Science"]

df["Average"] = df["Total"] / 3

print(df)
```

```python
print(df.sort_values("Total", ascending=False))
```

```python
##Top 3
print(df.sort_values("Total" , ascending=False).head(3))
```

**📊 Step 4 — Find Subject Averages**

```python
print("All Subject Averages")
print(df[['Math' , 'English' , 'Science']].mean())
```

Science subject has highest average among the 3 subjects.

**👩‍🎓 Step 5 — Gender Analysis**

```python
print("Average marks of male students")
males = df[df['Gender']=='Male']
print(males[["Math", "English", "Science", "Total", "Average"]].mean())
```

```python
print("Average marks of female students")
females = df[df['Gender']=='Female']
print(females[["Math", "English", "Science", "Total", "Average"]].mean())
```

Female students have higher Total Average Scores than Male students.

**📈 Step 6 — Visualization**

```python
import matplotlib.pyplot as plt

sub_avg = df[['Math' , 'English' , 'Science']].mean()

sub_avg.plot(kind="bar")

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

**🏆 Step 7 — Find the Best Student**

```python
best_performer = df.loc[df['Total'].idxmax()]
print("The top Performer is: ")
print(best_performer)
```

```python
low_performer = df.loc[df['Total'].idxmin()]
print("The low Performer is: ")
print(low_performer)
```

**Standard Deviation:**
Finding Most Variated Subject

```python
print("Standard Deviation of each subject:")

print(df[["Math", "English", "Science"]].std())

subject_variation = df[["Math", "English", "Science"]].std()

print("\nSubject with the most variation:")
print(subject_variation.idxmax())
```

std() means standard deviation.

Higher standard deviation → values are more spread out.

Lower standard deviation → values are closer to the average.

**Who is strongest and weakest in each subject?**

```python
subjects = ["Math" , "English" , "Science"]

for subject in subjects:
  highest = df.loc[df[subject].idxmax()]
  lowest = df.loc[df[subject].idxmin()]

  print(f"\n{subject}")
  print('Highest: ' , highest['Name'] , "-" , highest[subject])
  print('Lowest: ' , lowest['Name'] , "-" , lowest[subject])
```

**How many students are above/below the overall average?**

```python
overall_avg = df["Average"].mean()
print("Overall Average is : " , overall_avg)

above_avg = df[df["Average"] > overall_avg]
below_avg = df[df["Average"] < overall_avg]

print('\n')
print(len(above_avg) ," students are above overall average")
print(len(below_avg) , " students are below average")

print('\nAbove Average')
print(above_avg[['Name' , 'Average']])

print('\nBelow Average')
print(below_avg[['Name' , 'Average']])
```

**Performance Distribution**

A **histogram** is appropriate here because we want to see how many students fall within different Average-score ranges.

```python
plt.hist(df["Average"], bins=5)

plt.title("Distribution of Student Average Scores")
plt.xlabel("Average Score")
plt.ylabel("Number of Students")

plt.show()
```

**Gender Comparison Visualization**

```python
female_avg_score = females['Average'].mean()
male_avg_score = males['Average'].mean()

print("Average Females scores: " , female_avg_score)
print("Average Males scores: " , male_avg_score)
```

```python
gender_avg = df.groupby("Gender")["Average"].mean()

gender_avg.plot(kind='bar')
plt.title("Average Performance by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")

plt.show()
```

In this dataset, female students have a higher average performance than male students.

**Student Ranking Visualization**

```python
ranking = df.sort_values("Total" , ascending=False)
plt.bar(df['Name'] , df['Total'])
plt.title("Student Performance Ranking")
plt.xlabel("Student")
plt.ylabel("Total Marks")
plt.xticks(rotation=45)
plt.show()
```

**Correlation Between Subjects**

```python
print("Correlation Matrix")
correlation = df[['Math' , 'English' , 'Science']].corr()
print(correlation)
```
