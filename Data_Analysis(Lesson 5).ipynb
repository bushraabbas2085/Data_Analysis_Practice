```python
 import numpy as np
 marks = np.array([85, 90, 78, 92, 88])
 print(marks)
 print(np.max(marks))
 print(np.min(marks))
 print(np.mean(marks))
 print(np.sum(marks))
 print(marks+5)
```

```python
marks = np.array([78, 85, 92, 67, 88, 95, 73, 81, 90, 84])
# A
print(marks[-3:])
# B
print(marks[marks > 85])
# C
print(np.sum(marks > 85))
```

```python
arr=np.array([[1,2],
              [3,4],
              [5,6] ])
print(arr.shape)
print(arr[0])
print(arr[-1])
##Marks Array
marks = np.array([
    [78, 85, 92],
    [67, 88, 95],
    [73, 81, 90],
    [91, 76, 85]
])
print(marks[:,1])
print(marks[1,2])
print(np.mean(marks[:, 0]))
##Avg of each row
print(np.mean(marks, axis=1))
##vg of each column
print(np.mean(marks, axis=0))
print(np.max(marks))
```

```python
import numpy as np
marks = np.array([
    [78, 85, 92],
    [67, 88, 95],
    [73, 81, 90],
    [91, 76, 85]
])
print(marks[marks > 85])
print(np.sum(marks > 85))
print(marks[marks < 80])
print(np.max(marks[: , 0]))
print(np.min(marks[: , 1]))
```

```python
##Mini Project
print("Average of each Student/Row")
Avg = np.mean(marks , axis=1)
print(Avg)
print(np.max(Avg) , "is the highest average which is of Student 1")
print("\nClass Average")
print(np.mean(marks))
print("\nAverage of each Subject/Colmn")
Average = np.mean(marks , axis=0)
print(Average)
print(np.max(Average) , "is the highest average which is of Subject 3")
print("\nNo of Students who scored above 80 : " , np.sum(marks>80))
print("\nStudent 3 has least average among the 4 Students")
```
